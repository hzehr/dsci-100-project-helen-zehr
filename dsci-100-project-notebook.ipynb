#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

- Use any functions from the `tidyverse` to create **one** table that outputs the mean, minimum, and maximum of **all** numeric columns in your data, dropping the missing values if they exist.
- Show the outputted table.

</div>

```r
basketball_mean <- tidy_basketball_data |>
                summarize(across(conf_wins_pct:reg_wins_pct, ~ mean(.x, na.rm = TRUE))) #calculate the means for numeric colums dropping missing values
                    
basketball_min <- tidy_basketball_data |>
                summarize(across(conf_wins_pct:reg_wins_pct, ~ min(.x, na.rm = TRUE))) #calculate the mins for numeric colums dropping missing values                          
                    
basketball_max <- tidy_basketball_data |>
                summarize(across(conf_wins_pct:reg_wins_pct, ~ max(.x, na.rm = TRUE))) #calculate the max for numeric columns dropping missing values                       

column_1 <- c("mean", "minimum", "maximum") #create the summary statistic row names

summary_table <- bind_rows(basketball_mean, basketball_min, basketball_max)|> #combine the rows for mean, min, and max to create a table              
                 mutate(statistic=column_1, .before=1) #add a column containing the summary statistic row names and place it first *source used*

summary_table
```

Visualization 1 compares each category of tournament finish against the mean conference wins percentage for that group. This is relevant to my analysis because my research question was "Can conference win percentage and regional win percentage be used to predict the category of tournament finish?". In this plot I am able to visualize the relationship between conference win percentage and tournament finish. I can use this information to detect if there are any noticeable patterns that may help answer this question. However, this graph cannot answer the question on its own because it has no information about regional win percentage.

```r
options(repr.plot.height=8, repr.plot.width=12)#change size of plot so title fits above it
visualization_1 <-tidy_basketball_data|>
                    group_by(tourney_finish)|>
                    summarize(mean_conf_wins_pct= mean(conf_wins_pct, na.rm=TRUE))|> #use group_by() and summarize() to get the mean for each group in the categorical varaible
                     mutate(tourney_finish = fct_recode(tourney_finish, 
                    "Second Round Loss" = "second_round_loss",
                    "Opening Round Loss" = "opening_round_loss",
                    "Top 4 Loss" = "top_4_loss",
                    "Champ" = "champ",
                    "Top 8 Loss" = "top_8_loss",
                    "Top 2 Loss" = "top_2_loss",
                    "First Round Loss" = "first_round_loss",
                    "Top 16 Loss" = "top_16_loss"))|> #rename of the groups for the tournemnt finish categorical variable so they are human-readable
                    ggplot(aes(y = tourney_finish, x= mean_conf_wins_pct))+ #plot tournament finish against mean conference wins percentage, switch the x and y axis so the labels fit
                    geom_bar(stat= "identity")+ #create a bar chart where the x-value is the mean conference wins percenatge
                    labs(title= "Mean Conference Wins Percentage for Tourmanet Finishes",
                            x="Mean Conference Wins Percentage",
                            y= "Tournament Finish")+ #Human-readable titles 
                    theme(text=element_text(size=20)) #change the text size

visualization_1
```

Visualization 2 compares the regional win percentage against the conference win percentage and the different tourament finish categories are colour coded. This is relevant to answering my research question because I am not only able to see the relationship between regional win percentage and conference win percentage, but I am also able to search for patterns between the 2 variables and the final tournament finish category. This is more relevant than visualzation 1 because it includes both of the predictor variables (regional win percentage and conference win percentage) and their relationship with the variable I want to predict. 

```r
options(repr.plot.height=8, repr.plot.width=14)#change size of plot so title fits above it
visualization_2 <-tidy_basketball_data|>
                mutate(tourney_finish = fct_recode(tourney_finish, 
                    "Second Round Loss" = "second_round_loss",
                    "Opening Round Loss" = "opening_round_loss",
                    "Top 4 Loss" = "top_4_loss",
                    "Champ" = "champ",
                    "Top 8 Loss" = "top_8_loss",
                    "Top 2 Loss" = "top_2_loss",
                    "First Round Loss" = "first_round_loss",
                    "Top 16 Loss" = "top_16_loss"))|> #rename of the groups for the tournemnt finish categorical variable so they are human-readable
                ggplot(aes(x=reg_wins_pct, y=conf_wins_pct, color=tourney_finish))+ #plot regional win percentage against conference win percentage and colour the different tournament finish categories
                geom_point(alpha=0.6)+ #change alpha so it is easier to read
                scale_color_brewer(palette = "Set2")+ #modify the colour palate
                labs(title= "Conference Wins Percentage vs Regional Wins Percentage for Tournament Finishes",
                            y="Conference Wins Percentage",
                            x= "Regional Wins Percentage",
                            color="Tournament Finish")+ #Human-readable titles 
                    theme(text=element_text(size=20)) #change the text size

visualization_2
```
